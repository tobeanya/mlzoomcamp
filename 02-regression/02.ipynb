```python
# ml_imports.py
# Comprehensive import cell for ML projects.
# This file attempts imports and reports missing packages.
# Set install_missing=True to auto-install via pip (not recommended in all environments).

import sys
import subprocess
import importlib
import types
import random

# Configure behavior
INSTALL_MISSING = True  # set True to attempt pip install missing packages
SEED = 42

# List of import statements to attempt (as strings)
_import_lines = [
    "import numpy as np",
    "import pandas as pd",
    "import scipy as sp",
    "import matplotlib.pyplot as plt",
    "import seaborn as sns",
    "import sklearn",
    "from sklearn import model_selection, preprocessing, pipeline, ensemble, linear_model, metrics, impute, feature_selection",
    "import joblib",
    "import xgboost as xgb",
    "import lightgbm as lgb",
    "import catboost",
    "import tensorflow as tf",
    "import torch",
    "import torchvision",
    "from PIL import Image",
    "import cv2",
    "import transformers",
    "import tokenizers",
    "import optuna",
    "import shap",
    "import lime",
    "import imblearn",
    "import plotly",
    "import statsmodels.api as sm",
    "from tqdm import tqdm",
    "import requests",
    "from bs4 import BeautifulSoup",
]

_missing = []

def pip_install(package_name):
    "Install package via pip using the current Python interpreter."
    try:
        print(f"Installing {package_name} ...")
        subprocess.check_call([sys.executable, "-m", "pip", "install", package_name])
        return True
    except Exception as e:
        print(f"Failed to install {package_name}: {e}")
        return False

# Try imports
for line in _import_lines:
    try:
        exec(line, globals())
    except Exception as exc:
        pkg_name = line.split()[1] if line.startswith("import ") else line.split()[1]
        _missing.append((line, str(exc)))
        if INSTALL_MISSING:
            # derive pip package name heuristically from the import line
            pkg = pkg_name.split('.')[0].strip(',')
            pip_ok = pip_install(pkg)
            if pip_ok:
                try:
                    exec(line, globals())
                    # remove from missing if succeeded
                    _missing = [m for m in _missing if m[0] != line]
                except Exception:
                    pass

# Seed RNGs where possible
random.seed(SEED)
try:
    np.random.seed(SEED)
except Exception:
    pass

try:
    torch.manual_seed(SEED)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(SEED)
except Exception:
    pass

try:
    tf.random.set_seed(SEED)
except Exception:
    pass

# Summary
print("ML imports summary:")
if not _missing:
    print("  All requested imports succeeded.")
else:
    print(f"  {len(_missing)} import(s) failed. Examples:")
    for line, err in _missing[:10]:
        print(f"   - `{line}` -> {err}")

# Common convenience settings
try:
    sns.set_theme()
    plt.rcParams["figure.figsize"] = (10, 6)
except Exception:
    pass

# Export some common names for convenience in notebooks/scripts that import this module
__all__ = [
    "np", "pd", "sp", "plt", "sns", "sklearn", "model_selection", "preprocessing",
    "pipeline", "ensemble", "linear_model", "metrics", "impute", "feature_selection",
    "joblib", "xgb", "lgb", "catboost", "tf", "torch", "torchvision", "Image", "cv2",
    "transformers", "tokenizers", "optuna", "shap", "lime", "imblearn", "plotly",
    "sm", "tqdm", "requests", "BeautifulSoup", "SEED"
]
```
